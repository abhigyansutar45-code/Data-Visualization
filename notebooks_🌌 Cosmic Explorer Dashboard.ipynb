```python

import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
import ipywidgets as widgets

from IPython.display import display, HTML, clear_output
from datetime import datetime

# Reproducible sample data
np.random.seed(42)

# -----------------------------------------
# CELESTIAL OBJECT DATA
# -----------------------------------------

object_names = [
    "Andromeda Galaxy", "Orion Nebula", "Crab Nebula",
    "Sombrero Galaxy", "Whirlpool Galaxy", "Eagle Nebula",
    "Helix Nebula", "Triangulum Galaxy", "Carina Nebula",
    "Horsehead Nebula", "Ring Nebula", "Pinwheel Galaxy",
    "Centaurus A", "Lagoon Nebula", "Veil Nebula",
    "Tarantula Nebula", "Butterfly Nebula", "Sculptor Galaxy",
    "Dumbbell Nebula", "Omega Nebula"
]

object_types = [
    "Galaxy", "Nebula", "Nebula", "Galaxy", "Galaxy",
    "Nebula", "Nebula", "Galaxy", "Nebula", "Nebula",
    "Nebula", "Galaxy", "Galaxy", "Nebula", "Nebula",
    "Nebula", "Nebula", "Galaxy", "Nebula", "Nebula"
]

objects = pd.DataFrame({
    "Object": object_names,
    "Type": object_types,
    "Distance_LY": np.random.uniform(500, 3000000, 20),
    "Brightness": np.random.uniform(1, 100, 20),
    "Temperature_K": np.random.uniform(2500, 15000, 20),
    "Discovery_Year": np.random.randint(1700, 2025, 20),
    "Mass_Solar": np.random.uniform(0.1, 100000, 20),
    "X": np.random.uniform(-100, 100, 20),
    "Y": np.random.uniform(-100, 100, 20),
    "Z": np.random.uniform(-100, 100, 20)
})

# -----------------------------------------
# PLANET DATA
# -----------------------------------------

planets = pd.DataFrame({
    "Planet": [
        "Mercury", "Venus", "Earth", "Mars",
        "Jupiter", "Saturn", "Uranus", "Neptune"
    ],
    "Diameter_KM": [
        4879, 12104, 12742, 6779,
        139820, 116460, 50724, 49244
    ],
    "Distance_Million_KM": [
        57.9, 108.2, 149.6, 227.9,
        778.6, 1433.5, 2872.5, 4495.1
    ],
    "Moons": [0, 0, 1, 2, 95, 146, 28, 16],
    "Orbital_Period_Days": [
        88, 225, 365, 687, 4333, 10759, 30687, 60190
    ],
    "Gravity": [
        3.7, 8.87, 9.81, 3.71,
        24.79, 10.44, 8.69, 11.15
    ]
})

# -----------------------------------------
# SPACE MISSION DATA
# -----------------------------------------

mission_names = [
    "Voyager 1", "Voyager 2", "Pioneer 10",
    "Pioneer 11", "New Horizons", "Cassini",
    "Galileo", "Juno", "Mars Reconnaissance Orbiter",
    "James Webb Space Telescope", "Hubble Space Telescope",
    "Perseverance", "Curiosity", "OSIRIS-REx",
    "Chandrayaan-3", "Apollo 11", "Artemis I",
    "Rosetta", "Hayabusa2", "Dawn"
]

missions = pd.DataFrame({
    "Mission": mission_names,
    "Agency": np.random.choice(
        ["NASA", "ESA", "ISRO", "JAXA"],
        len(mission_names)
    ),
    "Launch_Year": np.random.randint(1960, 2025, len(mission_names)),
    "Mission_Duration_Years": np.random.uniform(
        0.5, 45, len(mission_names)
    ),
    "Mission_Category": np.random.choice(
        ["Planetary", "Lunar", "Deep Space", "Observatory"],
        len(mission_names)
    )
})

# -----------------------------------------
# ASTEROID DATA
# -----------------------------------------

asteroids = pd.DataFrame({
    "Asteroid_ID": [
        f"AST-{i:03d}" for i in range(1, 101)
    ],
    "Diameter_KM": np.random.uniform(0.05, 25, 100),
    "Distance_Million_KM": np.random.uniform(0.1, 50, 100),
    "Velocity_KM_S": np.random.uniform(5, 45, 100),
    "Orbit_Type": np.random.choice(
        ["Near Earth", "Main Belt", "Other"],
        100
    )
})

print("COSMIC EXPLORER DATA LOADED")
print("--------------------------------")
print("Celestial Objects:", len(objects))
print("Planets:", len(planets))
print("Space Missions:", len(missions))
print("Asteroids:", len(asteroids))
```

```python

# -----------------------------------------
# DASHBOARD HEADER
# -----------------------------------------

display(HTML("""
<div style="
    background:linear-gradient(135deg,#080A12,#171033,#101B35);
    padding:30px;
    border-radius:18px;
    color:white;
    border:1px solid #34305C;
">
    <h1 style="color:#A78BFA;font-size:34px;">
        🌌 COSMIC EXPLORER
    </h1>

    <h3 style="color:#00D9FF;">
        ADVANCED SPACE ANALYTICS DASHBOARD
    </h3>

    <p>
        Explore celestial objects, planetary systems,
        space missions, and cosmic discoveries.
    </p>

    <p style="color:#94A3B8;">
        Python Analytics • Interactive Visualization • Astronomy
    </p>
</div>
"""))

# -----------------------------------------
# SIDEBAR FILTERS
# -----------------------------------------

object_type_filter = widgets.Dropdown(
    options=["All"] + sorted(objects["Type"].unique().tolist()),
    value="All",
    description="Object Type:"
)

planet_filter = widgets.Dropdown(
    options=planets["Planet"].tolist(),
    value="Earth",
    description="Planet:"
)

mission_agency_filter = widgets.Dropdown(
    options=["All"] + sorted(missions["Agency"].unique().tolist()),
    value="All",
    description="Agency:"
)

search_box = widgets.Text(
    value="",
    placeholder="Search celestial objects...",
    description="Search:"
)

distance_slider = widgets.FloatRangeSlider(
    value=[
        float(objects["Distance_LY"].min()),
        float(objects["Distance_LY"].max())
    ],
    min=float(objects["Distance_LY"].min()),
    max=float(objects["Distance_LY"].max()),
    step=1000,
    description="Distance:",
    continuous_update=False,
    layout=widgets.Layout(width="95%")
)

dashboard_tabs = widgets.Tab()

dashboard_tabs.children = [
    widgets.Output(),
    widgets.Output(),
    widgets.Output(),
    widgets.Output()
]

dashboard_tabs.set_title(0, "🌌 Universe")
dashboard_tabs.set_title(1, "🪐 Planets")
dashboard_tabs.set_title(2, "🚀 Missions")
dashboard_tabs.set_title(3, "☄️ Asteroids")

filters = widgets.VBox([
    widgets.HTML("<h3>🔭 Explorer Controls</h3>"),
    search_box,
    object_type_filter,
    distance_slider,
    widgets.HTML("<hr><h3>🪐 Planet Explorer</h3>"),
    planet_filter,
    widgets.HTML("<hr><h3>🚀 Mission Explorer</h3>"),
    mission_agency_filter
])

display(filters)
display(dashboard_tabs)
```

```python

# -----------------------------------------
# DASHBOARD RENDERING
# -----------------------------------------

def render_dashboard(*args):

    # Filter celestial objects
    filtered = objects.copy()

    selected_type = object_type_filter.value
    search = search_box.value.lower()

    if selected_type != "All":
        filtered = filtered[
            filtered["Type"] == selected_type
        ]

    filtered = filtered[
        (filtered["Distance_LY"] >= distance_slider.value[0]) &
        (filtered["Distance_LY"] <= distance_slider.value[1])
    ]

    if search:
        filtered = filtered[
            filtered["Object"].str.lower().str.contains(
                search, regex=False
            )
        ]

    # Filter missions
    filtered_missions = missions.copy()

    selected_agency = mission_agency_filter.value

    if selected_agency != "All":
        filtered_missions = filtered_missions[
            filtered_missions["Agency"] == selected_agency
        ]

    # -------------------------------------
    # TAB 1: UNIVERSE
    # -------------------------------------

    with dashboard_tabs.children[0]:

        clear_output(wait=True)

        display(HTML("""
        <h2 style="color:#8B5CF6;">
            🌌 Universe Explorer
        </h2>
        """))

        # KPI cards
        display(HTML(f"""
        <div style="
            display:flex;
            flex-wrap:wrap;
            gap:12px;
        ">
            <div style="
                background:#101423;
                padding:18px;
                border-radius:12px;
                color:white;
                min-width:150px;
            ">
                <h3 style="color:#A78BFA;">
                    {len(filtered)}
                </h3>
                <p>Celestial Objects</p>
            </div>

            <div style="
                background:#101423;
                padding:18px;
                border-radius:12px;
                color:white;
                min-width:150px;
            ">
                <h3 style="color:#00D9FF;">
                    {len(filtered_missions)}
                </h3>
                <p>Space Missions</p>
            </div>

            <div style="
                background:#101423;
                padding:18px;
                border-radius:12px;
                color:white;
                min-width:150px;
            ">
                <h3 style="color:#22C55E;">
                    {len(planets)}
                </h3>
                <p>Planets</p>
            </div>
        </div>
        """))

        if not filtered.empty:

            # Cosmic scatter plot
            fig = px.scatter_3d(
                filtered,
                x="X",
                y="Y",
                z="Z",
                color="Type",
                size="Brightness",
                hover_name="Object",
                hover_data=[
                    "Distance_LY",
                    "Temperature_K",
                    "Discovery_Year"
                ],
                title="3D Cosmic Object Explorer"
            )

            fig.update_layout(
                template="plotly_dark",
                height=650,
                paper_bgcolor="#080A12",
                plot_bgcolor="#080A12"
            )

            fig.show()

            # Object distance chart
            distance_chart = px.bar(
                filtered.sort_values("Distance_LY"),
                x="Object",
                y="Distance_LY",
                color="Type",
                title="Celestial Object Distance"
            )

            distance_chart.update_layout(
                template="plotly_dark",
                xaxis_tickangle=-45,
                height=450
            )

            distance_chart.show()

            # Temperature chart
            temperature_chart = px.scatter(
                filtered,
                x="Distance_LY",
                y="Temperature_K",
                color="Type",
                size="Brightness",
                hover_name="Object",
                title="Cosmic Temperature vs Distance"
            )

            temperature_chart.update_layout(
                template="plotly_dark",
                height=450
            )

            temperature_chart.show()

            display(filtered)

        else:
            display(HTML("<h3>No matching objects found.</h3>"))

    # -------------------------------------
    # TAB 2: PLANETS
    # -------------------------------------

    with dashboard_tabs.children[1]:

        clear_output(wait=True)

        selected_planet = planets[
            planets["Planet"] == planet_filter.value
        ]

        display(HTML("""
        <h2 style="color:#00D9FF;">
            🪐 Planet Explorer
        </h2>
        """))

        display(selected_planet)

        fig = px.bar(
            planets,
            x="Planet",
            y="Diameter_KM",
            color="Planet",
            title="Planet Diameter Comparison"
        )

        fig.update_layout(
            template="plotly_dark",
            height=450
        )

        fig.show()

        fig = px.scatter(
            planets,
            x="Distance_Million_KM",
            y="Orbital_Period_Days",
            size="Diameter_KM",
            color="Planet",
            hover_name="Planet",
            title="Planet Distance vs Orbital Period"
        )

        fig.update_layout(
            template="plotly_dark",
            height=450
        )

        fig.show()

        fig = px.bar(
            planets,
            x="Planet",
            y="Moons",
            color="Planet",
            title="Known Moon Counts in Sample Dataset"
        )

        fig.update_layout(
            template="plotly_dark",
            height=400
        )

        fig.show()

    # -------------------------------------
    # TAB 3: SPACE MISSIONS
    # -------------------------------------

    with dashboard_tabs.children[2]:

        clear_output(wait=True)

        display(HTML("""
        <h2 style="color:#A78BFA;">
            🚀 Space Mission Analytics
        </h2>
        """))

        display(filtered_missions)

        fig = px.histogram(
            filtered_missions,
            x="Launch_Year",
            color="Agency",
            title="Space Missions by Launch Year"
        )

        fig.update_layout(
            template="plotly_dark",
            height=450
        )

        fig.show()

        fig = px.bar(
            filtered_missions,
            x="Mission",
            y="Mission_Duration_Years",
            color="Mission_Category",
            title="Mission Duration Comparison"
        )

        fig.update_layout(
            template="plotly_dark",
            xaxis_tickangle=-45,
            height=500
        )

        fig.show()

        fig = px.pie(
            filtered_missions,
            names="Agency",
            title="Mission Agency Distribution"
        )

        fig.update_layout(
            template="plotly_dark",
            height=450
        )

        fig.show()

    # -------------------------------------
    # TAB 4: ASTEROIDS
    # -------------------------------------

    with dashboard_tabs.children[3]:

        clear_output(wait=True)

        display(HTML("""
        <h2 style="color:#00D9FF;">
            ☄️ Asteroid Intelligence
        </h2>
        """))

        display(asteroids.head(20))

        fig = px.scatter(
            asteroids,
            x="Distance_Million_KM",
            y="Velocity_KM_S",
            size="Diameter_KM",
            color="Orbit_Type",
            hover_name="Asteroid_ID",
            title="Asteroid Distance vs Velocity"
        )

        fig.update_layout(
            template="plotly_dark",
            height=500
        )

        fig.show()

        fig = px.histogram(
            asteroids,
            x="Diameter_KM",
            color="Orbit_Type",
            title="Asteroid Diameter Distribution"
        )

        fig.update_layout(
            template="plotly_dark",
            height=450
        )

        fig.show()

        fig = px.box(
            asteroids,
            x="Orbit_Type",
            y="Velocity_KM_S",
            color="Orbit_Type",
            title="Asteroid Velocity Distribution"
        )

        fig.update_layout(
            template="plotly_dark",
            height=450
        )

        fig.show()


# -----------------------------------------
# CONNECT FILTERS
# -----------------------------------------

object_type_filter.observe(
    render_dashboard,
    names="value"
)

planet_filter.observe(
    render_dashboard,
    names="value"
)

mission_agency_filter.observe(
    render_dashboard,
    names="value"
)

search_box.observe(
    render_dashboard,
    names="value"
)

distance_slider.observe(
    render_dashboard,
    names="value"
)

# Initial rendering
render_dashboard()
```

```python

# Export astronomy datasets

objects.to_csv(
    "cosmic_objects.csv",
    index=False
)

planets.to_csv(
    "planet_data.csv",
    index=False
)

missions.to_csv(
    "space_missions.csv",
    index=False
)

asteroids.to_csv(
    "asteroid_data.csv",
    index=False
)

print("CSV exports completed!")
print("Files:")
print("- cosmic_objects.csv")
print("- planet_data.csv")
print("- space_missions.csv")
print("- asteroid_data.csv")
```
