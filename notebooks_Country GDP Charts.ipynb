```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

# Illustrative nominal GDP values (USD trillions), rounded for a classroom visualization.
# Values are approximate and should not be treated as a live/current economic dataset.
data = {
    "Country": ["United States", "China", "Germany", "Japan", "India",
                "United Kingdom", "France", "Italy", "Canada", "Brazil"],
    "GDP_USD_Trillion": [29.2, 18.7, 4.7, 4.1, 3.9, 3.6, 3.2, 2.4, 2.2, 2.2]
}
df = pd.DataFrame(data).sort_values("GDP_USD_Trillion", ascending=False)

fig, axes = plt.subplots(2, 2, figsize=(16, 11))
fig.suptitle("COUNTRY GDP DASHBOARD", fontsize=22, fontweight="bold", y=0.98)

# Bar chart: horizontal bars
ax = axes[0, 0]
bar_df = df.sort_values("GDP_USD_Trillion")
ax.barh(bar_df["Country"], bar_df["GDP_USD_Trillion"])
ax.set_title("1. Bar Chart — GDP by Country")
ax.set_xlabel("GDP (USD trillions)")
ax.grid(axis="x", alpha=0.25)
for i, v in enumerate(bar_df["GDP_USD_Trillion"]):
    ax.text(v + 0.25, i, f"{v:.1f}", va="center", fontsize=8)
ax.set_xlim(0, 33)

# Line chart: ranked countries (for comparison, not a time series)
ax = axes[0, 1]
ax.plot(df["Country"], df["GDP_USD_Trillion"], marker="o", linewidth=2)
ax.set_title("2. Line Chart — GDP Comparison")
ax.set_ylabel("GDP (USD trillions)")
ax.tick_params(axis="x", rotation=45, labelsize=8)
ax.grid(alpha=0.25)

# Pie chart: GDP share of these 10 countries
ax = axes[1, 0]
ax.pie(df["GDP_USD_Trillion"], labels=df["Country"], autopct="%1.1f%%",
       startangle=90, pctdistance=0.72, textprops={"fontsize": 8})
ax.set_title("3. Pie Chart — Share of Combined GDP")

# Column chart: vertical bars
ax = axes[1, 1]
ax.bar(df["Country"], df["GDP_USD_Trillion"])
ax.set_title("4. Column Chart — GDP by Country")
ax.set_ylabel("GDP (USD trillions)")
ax.tick_params(axis="x", rotation=45, labelsize=8)
ax.grid(axis="y", alpha=0.25)
for i, v in enumerate(df["GDP_USD_Trillion"]):
    ax.text(i, v + 0.35, f"{v:.1f}", ha="center", fontsize=8)
ax.set_ylim(0, 33)

fig.text(0.5, 0.015, "Illustrative rounded nominal GDP values • USD trillions • Not live/current data",
         ha="center", fontsize=10, color="dimgray")
plt.tight_layout(rect=[0, 0.04, 1, 0.95])

out = "/mnt/data/country_gdp_dashboard.png"
csv_out = "/mnt/data/country_gdp_data.csv"
fig.savefig(out, dpi=160, bbox_inches="tight")
df.to_csv(csv_out, index=False)
plt.show()

print("Files created:")
print(out)
print(csv_out)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Country": ["United States", "China", "Germany", "Japan", "India",
                "United Kingdom", "France", "Italy", "Canada", "Brazil"],
    "GDP_USD_Trillion": [29.2, 18.7, 4.7, 4.1, 3.9, 3.6, 3.2, 2.4, 2.2, 2.2]
}).sort_values("GDP_USD_Trillion", ascending=False)

fig, axes = plt.subplots(2, 2, figsize=(15, 10))
fig.suptitle("COUNTRY GDP DASHBOARD", fontsize=21, fontweight="bold")

# 1. Horizontal bar chart
bar_df = df.sort_values("GDP_USD_Trillion")
axes[0, 0].barh(bar_df["Country"], bar_df["GDP_USD_Trillion"])
axes[0, 0].set_title("1. Bar Chart — GDP by Country")
axes[0, 0].set_xlabel("GDP (USD trillions)")
axes[0, 0].grid(axis="x", alpha=0.25)

# 2. Line chart (country comparison, not a time series)
axes[0, 1].plot(df["Country"], df["GDP_USD_Trillion"], marker="o", linewidth=2)
axes[0, 1].set_title("2. Line Chart — GDP Comparison")
axes[0, 1].set_ylabel("GDP (USD trillions)")
axes[0, 1].tick_params(axis="x", rotation=45, labelsize=8)
axes[0, 1].grid(alpha=0.25)

# 3. Pie chart (interpreting “hie chart” as “pie chart”)
axes[1, 0].pie(df["GDP_USD_Trillion"], labels=df["Country"], autopct="%1.1f%%",
               startangle=90, textprops={"fontsize": 8})
axes[1, 0].set_title("3. Pie Chart — Share of Combined GDP")

# 4. Column chart
axes[1, 1].bar(df["Country"], df["GDP_USD_Trillion"])
axes[1, 1].set_title("4. Column Chart — GDP by Country")
axes[1, 1].set_ylabel("GDP (USD trillions)")
axes[1, 1].tick_params(axis="x", rotation=45, labelsize=8)
axes[1, 1].grid(axis="y", alpha=0.25)

fig.text(0.5, 0.01, "Illustrative rounded nominal GDP values • USD trillions • Not live/current data",
         ha="center", fontsize=9)
plt.tight_layout(rect=[0, 0.04, 1, 0.94])

image_path = "/mnt/data/country_gdp_dashboard.png"
csv_path = "/mnt/data/country_gdp_data.csv"
fig.savefig(image_path, dpi=150, bbox_inches="tight")
df.to_csv(csv_path, index=False)
plt.show()
print("Dashboard image:", image_path)
print("Country GDP data CSV:", csv_path)
```
